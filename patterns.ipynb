```python
def right_traingle(num):
    for i in range(1,num+1):
        print(" "*(num-1), end ="")
        for j in range(1,i+1):
            #print(" "*, end ="")
            print("*", end ="")
        print()

right_traingle(5)
    
```

```python
def right_traingle_mirror(num):
    for i in range(1,num+1):
        print(" "*(num-i), end ="")
        for j in range(1,i+1):
            print("*", end ="")
        print()
right_traingle_mirror(5)
```

```python
def pyramid(num):
    for i in range(1, num+1):
        print(" "*(num-i), end ="")
        for j in range(1, i+1):
            print(" *", end ="")
        print()
pyramid(5)
```

```python
def hollow_square(num):
    for i in range(1, num+1):
        if i == 1 or i == num:
            for j in range(1, num+1):
                print(" *", end="")
        else:
            for j in range(1, num+1):
                if j == 1 or j == num:
                    print(" *", end="")
                else:
                    print("  ", end ="")

        print()
hollow_square(5)
```

```python
def hollow_triangle(num):
    for i in range(1,num+2):
        if i == 1 or i ==2 or i == (num+1):
            for j in range(1,i+1):
                print(" *", end ="")
        else:
            for k in range(1, i+1):
                if k == 1 or k == i:
                    print(" *", end ="")
                else:
                    print("  ", end ="")
        print()
hollow_triangle(7)
```

```python
def hallow_diamond(num):
    for i in range(1, num+1):
        for j in range(num-i, 0, -1):
            print(" ", end ="")
        for k in range(1, i+1):
            if  k == 1 or k == i:
                print("* ", end ="")
            else: 
                print("  ", end ="")
        print()
    for i in range(1, num):
        for j in range(1, i+1):
            print(" ", end ="")
        for k in range(num-i, 0, -1):
            if k == 1 or k == (num-i):
                print("* ", end ="")
            else:
                print("  ", end ="")
        print()
hallow_diamond(5)
```

```python
def number_traingle(num):
    for i in range(1, num+1):
        for j in range(1,i+1):
            print(j, end= "")
        print()
number_traingle(5)
```

```python
def mirror_number_traingle(num):
    for i in range(1, num+1):
        for j in range(num - i, 0, -1):
            print(" ", end ="")
        for k in range(1, i+1):
            print(k, end="")
        print()
mirror_number_traingle(5)
```

```python
def inverted_number_traingle(num):
    for i in range(num, 0, -1):
        for j in range(1, i+1):
            print(j, end ="")
        print()
inverted_number_traingle(5)
```

```python
def number_pyramid(num):
    for i in range(1, num+1):
        for j in range(num-i):
            print(" ", end ="")
        for k in range(1, i+1):
            print(k, end ="")
        for l in range(i-1,0,-1):
            print(l, end ="")
        print()
number_pyramid(7)
```

```python
def inverted_number_pyramid(num):
    for i in range(1,num+1):
        for k in range(i):
            print(" ", end ="")
        for j in range(1,num+2-i):
            print(j, end ="")
        for l in range(num-i, 0, -1):
            print(l, end="")
        print()
inverted_number_pyramid(7)

```

```python
# here when we use max() funct in a list of strings it will not check the len of each strings
# instead it will check the 1st character alphabetically and return the largest (z>a) alphabetical string
# what if 2 string have a same 1st aphabetical character?
#  then it will check the 2nd character and so on until it finds the largest string alphabetically

city = ["New York", "Phoenix" ,"LosAngeles", "Chicago", "Houston"]
print(max(city))
print(max(city, key =len)) # when we use key = len it will return the largest string based on the length of the string
```
